```python

# Faturamento total
df["SALES"].sum()

# Por ano
df.groupby("YEAR_ID")["SALES"].sum()

# Por país
df.groupby("COUNTRY")["SALES"].sum().sort_values(ascending=False).head(10)

# Por produto
df.groupby("PRODUCTLINE")["SALES"].sum()
```

```python
import matplotlib.pyplot as plt

vendas_produto = df.groupby("PRODUCTLINE")["SALES"].sum().sort_values()

vendas_produto.plot(kind="barh", figsize=(10,6))
plt.title("Faturamento por Linha de Produto")
plt.xlabel("Vendas")
plt.ylabel("Produto")
plt.show()
```

```python
df.dtypes
```

```python
df['Ano'] = df['ORDERDATE'].dt.year
df['Mes'] = df['ORDERDATE'].dt.month
df['Mes_nome'] = df['ORDERDATE'].dt.month_name()
df[['ORDERDATE', 'Ano', 'Mes', 'Mes_nome']].head(10)
```

```python
#Qual ano vendu mais?

vendas_ano = df.groupby('Ano')['SALES'].sum().sort_index()
vendas_ano


```

```python
# Gráfico de linha para evolução de vendas por ano

plt.Figure(figsize=(8,5))
plt.plot(vendas_ano.index, vendas_ano.values, marker= 'o')

plt.title('Evolução de Vendas por Ano')
plt.xlabel('Ano')
plt.ylabel('vendas')
plt.grid(True)

plt.show()
```

```python
#Quais mêses vendem mais?

vendas_mes = df.groupby('Mes')['SALES'].sum()
plt.figure(figsize=(10,5))
plt.bar(vendas_mes.index, vendas_mes.values)

plt.title('Vendas por Mês')
plt.xlabel('Mês')
plt.ylabel('Vendas')

plt.show()


```

## Análise Temporal

- Observa-se variação significativa nas vendas ao longo dos anos, com crescimento em 2004 e queda em 2005.
- Alguns meses concentram maior volume de vendas, especialmente maio, outubro e novembro.
- Esse padrão sugere sazonalidade no negócio.
- Possivelmente, nesses meses ocorrem campanhas promocionais, lançamentos ou maior demanda do mercado.
- Recomenda-se analisar as estratégias comerciais e de pós-venda aplicadas nesses períodos para replicação ao longo do ano.
