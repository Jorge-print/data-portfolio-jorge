```python
# Faturamento por cliente

receita_cliente = (df.groupby("CUSTOMERNAME")["SALES"]
                   .sum()
                   .sort_values(ascending=False)
                   .reset_index())

receita_cliente.head(92)

```

```python
#Percentual acumulado dos clientes

receita_acumulada = receita_cliente["SALES"].cumsum()
percentual_acumulado = receita_acumulada / receita_cliente["SALES"].sum() * 100

pareto_df = pd.DataFrame({
    "Receita": receita_cliente['SALES'],
    "Percentual_Acumulado": percentual_acumulado
})

pareto_df.head()

```

```python
clientes_80 = pareto_df[pareto_df["Percentual_Acumulado"] <= 80]

quantidade_clientes_80 = clientes_80.shape[0]
total_clientes = receita_cliente.shape[0]

print(f"{quantidade_clientes_80} clientes geram aproximadamente 80% da receita.")
print(f"Total de clientes: {total_clientes}")
```

```python
plt.figure(figsize=(10,6))

plt.plot(percentual_acumulado.values)
plt.axhline(y=80, color="g", linestyle="--")

plt.title("Curva de Pareto - Clientes")
plt.xlabel("Clientes")
plt.ylabel("Percentual acumulado da receita")

plt.show()
```

```python
receita_cliente = receita_cliente.sort_values(by="SALES", ascending=False)

receita_total = receita_cliente["SALES"].sum()

receita_cliente["perc_receita"] = receita_cliente["SALES"] / receita_total

receita_cliente["perc_acumulado"] = receita_cliente["perc_receita"].cumsum()

```

```python
def classificar(x):
    if x <= 0.5:
        return "Alto Valor (A)"
    elif x <= 0.8:
        return "Médio Valor (B)"
    else:
        return "Baixo Valor (C)"

receita_cliente["segmento"] = receita_cliente["perc_acumulado"].apply(classificar)
```
